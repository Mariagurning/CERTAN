```python
import numpy as np
import matplotlib.pyplot as plt

def remove_outliers_sequentially(df2, columns):
    df_cleaned = df2.copy()
    for col in columns:
        q1, q3 = np.percentile(df_cleaned[col], [25, 75])
        iqr = q3 - q1
        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)
        df_cleaned = df_cleaned[(df_cleaned[col] >= lower_bound) & (df_cleaned[col] <= upper_bound)]
    return df_cleaned

# Terapkan fungsi untuk menghapus outliers dari "Height" dan "Weight"
clean_data = remove_outliers_sequentially(df2, ['Height', 'Weight'])

# Plot boxplot setelah menghapus outliers
columns_to_plot = ['Weight', 'Height']
columns_to_plot = [col for col in columns_to_plot if col in clean_data.columns]

fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(clean_data[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

# Analisa Missing Value - Titanic Dataset

Berdasarkan pengecekan `df.isnull().sum()`, ditemukan missing value pada beberapa kolom:

- Age: memiliki missing value dalam jumlah cukup signifikan. Karena Age bersifat numerik dan datanya kemungkinan tidak simetris, digunakan median untuk mengisi nilai kosong, karena median lebih tahan terhadap outlier dibanding mean.

- Embarked: hanya sedikit baris yang kosong. Karena kolom ini bersifat kategorikal (pelabuhan keberangkatan), digunakan modus (nilai yang paling sering muncul) untuk  mengisi nilai kosong, sesuai kesepakatan umum dalam menangani data kategorikal.

- Cabin: memiliki persentase missing value yang sangat tinggi (lebih dari 70% data kosong). Karena terlalu banyak yang hilang, mengisi nilainya justru berisiko menambah bias pada analisis. Oleh karena itu, kolom ini dihapus (drop) daripada diisi.

Setelah proses ini dilakukan, hasil `df.isnull().sum()` menunjukkan seluruh kolom sudah bernilai 0, artinya tidak ada lagi missing value pada dataset.

- Bebas menggunakan matplotlib atau seaborn


```python
import numpy as np
import matplotlib.pyplot as plt

path = "diabetes.csv"
df_diabetes = pd.read_csv(path)
df_diabetes.head()
```

```python
# Pilih 3 kolom untuk deteksi outlier
columns_to_check = ['Glucose', 'BloodPressure', 'BMI']

# Visualisasi sebelum dibersihkan
fig, axs = plt.subplots(len(columns_to_check), 1, dpi=95, figsize=(7, 4 * len(columns_to_check)))
for i, col in enumerate(columns_to_check):
    axs[i].boxplot(df_diabetes[col], vert=False)
    axs[i].set_ylabel(col)
plt.suptitle("Sebelum Handling Outliers")
plt.show()
```

```python
def remove_outliers_sequentially(df, columns):
    df_cleaned = df.copy()
    for col in columns:
        q1, q3 = np.percentile(df_cleaned[col], [25, 75])
        iqr = q3 - q1
        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)
        df_cleaned = df_cleaned[(df_cleaned[col] >= lower_bound) & (df_cleaned[col] <= upper_bound)]
    return df_cleaned

clean_diabetes = remove_outliers_sequentially(df_diabetes, columns_to_check)
```

```python
# Visualisasi setelah dibersihkan
fig, axs = plt.subplots(len(columns_to_check), 1, dpi=95, figsize=(7, 4 * len(columns_to_check)))
for i, col in enumerate(columns_to_check):
    axs[i].boxplot(clean_diabetes[col], vert=False)
    axs[i].set_ylabel(col)
plt.suptitle("Setelah Handling Outliers")
plt.show()
```

# ### Analisa Outliers - Diabetes Dataset

Deteksi outlier dilakukan pada 3 kolom numerik: **Glucose**, **BloodPressure**, dan **BMI**, 
menggunakan metode Interquartile Range (IQR).

Sebelum dibersihkan:
Boxplot awal menunjukkan adanya titik-titik di luar whisker pada ketiga kolom, terutama pada BloodPressure dan BMI. Ini kemungkinan berasal dari nilai 0 yang secara medis tidak masuk akal  (misalnya tekanan darah 0), yang biasanya merupakan representasi data yang tidak tercatat, bukan nilai asli.

Proses pembersihan:
Outlier dihapus dengan menghitung Q1, Q3, dan IQR untuk masing-masing kolom, lalu membuang  baris yang nilainya berada di luar rentang [Q1 - 1.5*IQR, Q3 + 1.5*IQR]. Proses ini dilakukan  secara berurutan (sequentially) pada ketiga kolom menggunakan fungsi "remove_outliers_sequentially".

- Setelah dibersihkan: Boxplot menunjukkan rentang data yang lebih rapat pada ketiga kolom, dengan sebagian besar outlier ekstremsudah terbuang. Pada BloodPressure, masih tersisa 1-2 titik di luar whisker.
ini terjadi karena batas IQR dihitung ulang dari data yang sudah lebih rapat setelah 
pembersihan, sehingga rentang "wajar" menjadi lebih sempit dari sebelumnya. Untuk pembersihan 
yang lebih menyeluruh, proses ini bisa diulang beberapa kali (iteratif) hingga tidak ada lagi  outlier yang terdeteksi, namun untuk kebutuhan tugas ini satu kali proses sudah cukup merepresentasikan penerapan metode IQR.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, roc_curve, auc, roc_auc_score
from sklearn.preprocessing import label_binarize
from sklearn.multiclass import OneVsRestClassifier
# Load dataset
df = sns.load_dataset("iris")
# 1. Histogram → Distribusi
plt.figure(figsize=(6,4))
sns.histplot(df['petal_length'], bins=20, kde=True)
plt.title("Histogram: Distribution of Petal Length")
plt.show()
```

```python
# 2. Bar Chart → Count per category
plt.figure(figsize=(6,4))
sns.countplot(x="species", data=df)
plt.title("Bar Chart: Count of Species")
plt.show()
```

```python
# 3. Box Plot → Spread + Outlier detection
plt.figure(figsize=(6,4))
sns.boxplot(x="species", y="petal_length", data=df)
plt.title("Box Plot: Petal Length by Species")
plt.show()
```

```python
# 4. Scatter Plot → Relationship
plt.figure(figsize=(6,4))  
sns.scatterplot(x="sepal_length", y="petal_length", hue="species", data=df)
plt.title("Scatter Plot: Sepal Length vs Petal Length")
plt.show()
```

```python
# 5. Pair Plot → Multivariate relationship
sns.pairplot(df, hue="species")
plt.suptitle("Pair Plot: Feature Relationships", y=1.02)
plt.show()
```

```python
# 6. Line Plot → Example time series (simulate cumulative sum)
plt.figure(figsize=(6,4))
df["random_growth"] = np.random.randn(len(df)).cumsum()
sns.lineplot(x=df.index, y="random_growth", data=df)
plt.title("Line Plot: Simulated Growth Over Index")
plt.show()
```

```python
# 7. Heatmap → Correlation matrix
plt.figure(figsize=(6,4))
sns.heatmap(df.select_dtypes(include="number").corr(), annot=True, cmap="coolwarm")
plt.title("Heatmap: Feature Correlation")
plt.show()
```

```python
fig = plt.figure(figsize=(7,5))
ax = fig.add_subplot(111, projection="3d")

species = df["species"].unique()
colors = ["r", "g", "b"]

for sp, col in zip(species, colors):
    subset = df[df["species"] == sp]
    ax.scatter(subset["sepal_length"], subset["sepal_width"], subset["petal_length"], c=col, label=sp, s=50)

ax.set_xlabel("Sepal Length")
ax.set_ylabel("Sepal Width")
ax.set_zlabel("Petal Length")
ax.set_title("3D Scatter Plot: Sepal Length, Sepal Width, Petal Length")
ax.legend()
plt.show()
```

```python
# Split data
X = df.drop("species", axis=1)
y = df["species"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# 9.1. Confusion Matrix
cm = confusion_matrix(y_test, y_pred, labels=model.classes_)
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=model.classes_, yticklabels=model.classes_)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
# 9.2. ROC–AUC Curve (multi-class with One-vs-Rest)

# Binarize labels
y_bin = label_binarize(y, classes=model.classes_)
n_classes = y_bin.shape[1]

# Train OvR classifier
ovr_model = OneVsRestClassifier(RandomForestClassifier(n_estimators=100, random_state=42))
ovr_model.fit(X_train, label_binarize(y_train, classes=model.classes_))
y_score = ovr_model.predict_proba(X_test)
y_test_bin = label_binarize(y_test, classes=model.classes_)

# Plot ROC curve
plt.figure(figsize=(7,5))
for i in range(n_classes):
    fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_score[:, i])
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, label=f"Class {model.classes_[i]} (AUC = {roc_auc:.2f})")

plt.plot([0,1], [0,1], "k--")  # diagonal
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC–AUC Curve (One-vs-Rest)")
plt.legend()
plt.show()
```

```python
# 9.3. Feature Importance Plot → Random Forest
X = df.drop("species", axis=1)
y = df["species"]
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)
importances = model.feature_importances_
features = X.columns
plt.figure(figsize=(6,4))
sns.barplot(x=importances, y=features)
plt.title("Feature Importance Plot (Random Forest)")
plt.show()
```
