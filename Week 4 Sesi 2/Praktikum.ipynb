```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

Cell ini mengimport NumPy (untuk komputasi numerik dan array), Pandas (untuk mengolah data dalam bentuk DataFrame), dan fungsi load_iris dari Scikit-Learn. Baris iris = load_iris() memuat dataset Iris yang berisi 150 data bunga dengan 4 fitur numerik dan 3 kelas spesies.


```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

Data Iris diubah menjadi DataFrame dengan nama kolom sesuai feature_names. Kolom "Species" ditambahkan dengan mengubah label angka (iris.target) menjadi nama spesies lewat iris.target_names. Setelah itu data dipisah menjadi X (4 fitur: sepal length, sepal width, petal length, petal width) dan y (label spesies). Output print(X) menunjukkan 150 baris × 4 kolom.


```python
y
```

Menampilkan isi y, yaitu label spesies setiap data. Data tersusun berurutan per kelas: indeks 0–49 setosa, 50–99 versicolor, dan 100–149 virginica. Totalnya 150 label bertipe teks (object), sehingga perlu diubah menjadi angka.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

LabelEncoder mengubah label teks menjadi angka berdasarkan urutan abjad: setosa = 0, versicolor = 1, virginica = 2. Hasilnya array berisi 50 angka untuk masing-masing kelas. Objek lab disimpan agar nanti hasil prediksi berupa angka bisa dikembalikan ke nama spesies dengan inverse_transform.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Data dibagi dengan test_size=0.2, yaitu 80% (120 data) untuk melatih model dan 20% (30 data) untuk menguji model. random_state=0 membuat pengacakan selalu sama setiap kali dijalankan. Karena X dan y dibagi dengan random_state yang sama, pasangan fitur dan labelnya tetap sesuai.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```


GaussianNB diimpor lalu dilatih dengan fit(X_train, y_train). Berbeda dengan KNN yang hanya menyimpan data, di tahap ini Naïve Bayes benar-benar "belajar". Model menghitung prior P(C) setiap kelas, serta rata-rata (μ) dan standar deviasi (σ) setiap fitur di setiap kelas. Nilai-nilai inilah yang nanti dipakai di rumus Gaussian untuk menghitung P(X|C).

Model memprediksi 30 data uji. Untuk setiap data, model menghitung probabilitas posterior tiap kelas dengan mengalikan prior P(C) dengan P(X|C) dari keempat fitur (asumsi fitur saling independen), lalu memilih kelas dengan probabilitas tertinggi. Hasilnya array 30 prediksi dalam bentuk angka 0/1/2.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

accuracy_score membandingkan hasil prediksi dengan label sebenarnya, lalu dikali 100 agar tampil dalam persen. Akurasinya 96,67%, artinya 29 dari 30 data uji benar dan hanya 1 yang salah. Nilai ini sama dengan akurasi KNN (K = 3) pada praktikum sebelumnya, sehingga kedua metode sama-sama bekerja sangat baik pada dataset Iris.
